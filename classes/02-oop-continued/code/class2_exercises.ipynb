## 🟢 Core - Coffee Shop ☕
Create 2 classes - a CoffeeShop class and a CoffeeOrder class

**CoffeeOrder**
1. Instance attributes 
- customer_name, 
- drink_name, 
- size ("small", "medium", "large"),
- price.
  
1. Instance method

**CoffeeShop**
1. Instance attributes
- shop_name, 
- orders (list to store CoffeeOrder objects)
2. Instance methods
- add_order(order) - adds a CoffeeOrder to the orders list
- get_total_orders() - returns how many orders the shop has
- list_all_orders() - prints all order summaries

```python
from __future__ import annotations
```

```python
class CoffeeOrder:
    def __init__(self, customer_name: str, drink_name: str, size: str, price: float):
        # TODO: Add validation for size to ensure it's one of "Small", "Medium", or "Large"
        self.customer_name = customer_name
        self.drink_name = drink_name
        self.size = size
        self.price = price

    def get_order_summary(self) -> str:
        # TODO: Return a string summarizing the order in the format:
        # Silvia ordered a medium flat white for 32dkk"
        return f"{self.customer_name} ordered a {self.size} {self.drink_name} for {self.price:.2f}dkk"
```

```python
class CoffeeShop:
    def __init__(self, shop_name: str):
        # TODO: Initialize the CoffeeShop with a name and an empty list of orders
        self.shop_name = shop_name
        self.orders: list[CoffeeOrder] = []
    
    def add_order(self, order: CoffeeOrder) -> None:
        # TODO: Add a CoffeeOrder to the list of orders
        self.orders.append(order)
        
    def get_total_orders(self) -> int:
        # TODO: Return the total number of orders in the shop
        return len(self.orders)

    def list_all_orders(self):
        # TODO: Print the summary of all orders in the shop
        print ("\n".join(order.get_order_summary() for order in self.orders))

    def get_revenue(self) -> float:
        return sum(order.get_total_price() for order in self.orders)
```

```python
# create some coffee orders
c1 = CoffeeOrder("Silvia", "Latte", "Medium", 35.0)
c2 = CoffeeOrder("Julia", "Cappuccino", "Large", 47.0)

c3 = CoffeeOrder("Paul", "Chai Latte", "Small", 37.0)
c4 = CoffeeOrder("Jari", "Chocolate", "Large", 42.0)

```

```python
# summary of some orders
for coffee_order in [c1, c4]:
    print(coffee_order.get_order_summary())
```

## 🟢 Core — Exercise 1: The `Book` class

Build a `Book` class with:

**Instance attributes**
- `title` → **public**
- `author` → **public**
- `price` → **private**
- `stock` → **protected**

**Instance methods**
- `get_price()` — returns the price
- `get_stock()` — returns the stock
- `sell()` — decreases stock by 1, but only if `stock > 0`. Return `True` if the sale went
  through, `False` otherwise.
- `__str__` — e.g. `"Dune by Frank Herbert - 150 dkk (3 in stock)"`
- `__eq__` and `__lt__` — compare two books **by price**


```python
class Book:
    def __init__(self, title: str, author: str, price: float, stock: int):
        self.title = title          # public
        self.author = author        # public
        self.__price = price# TODO: store price as a PRIVATE attribute
        self._stock = stock# TODO: store stock as a PROTECTED attribute

    def get_price(self) -> float:
        # TODO
        return self.__price

    def get_stock(self) -> int:
        # TODO
        return self._stock

    def sell(self) -> bool:
        # TODO: decrease stock by 1 only if stock > 0, return True/False
        if self._stock > 0:
            self._stock -= 1
            return True
        return False

    def __str__(self) -> str:
        # TODO
        return f"{self.title} by {self.author} - {self.__price:.2f} dkk ({self._stock} in stock)"

    def __repr__(self) -> str:
        return f"Book(title={self.title}, author={self.author}, price={self.__price}, stock={self._stock})"

    def __eq__(self, other: object) -> bool:
        # TODO: equal if prices are equal
        if not isinstance (other, Book):
            raise TypeError(f"other object should be of type 'Book', encountered {type(other)}")
        return self.__price == other.__price


    def __lt__(self, other: Book) -> bool:
        # TODO: less-than based on price
        return self.__price < other.__price
```

Create an `EBook` class that **inherits from `Book`**:

- Uses `super().__init__(...)` to set up the inherited attributes.
- Overrides `__str__` to reuse the parent's `__str__` (via `super()`) and append something like

```python
class EBook(Book):
    def __init__(self, title: str, author: str, price: float, stock: int, file_size_mb: float):
        # TODO: call the parent constructor with super()
        super().__init__(title, author, price, stock)
        # TODO: store file_size_mb
        self.file_size_mb = file_size_mb

    def __str__(self) -> str:
        # TODO: reuse Book's __str__ with super() and add the ebook info
        return super().__str__() + f" [ebook, {self.file_size_mb:.1f}MB]"
```

## 🟡 Stretch — Exercise 3: `Library` class (Core + Stretch methods)

Mixing regular methods (add_book, total_value, find_by_author) with __len__ and __getitem__ ("mix dunders and regular methods").

Create a `Library` class that **stores** `Book` objects (composition, not inheritance).

- `__init__(self, name)` 🟡 — stores the library's name and an empty list of books
- `add_book(book)` 🟡 — adds a book to the library
- `total_value()` 🟡 — returns the sum of `price * stock` over all books
- `find_by_author(author)` 🟡 — returns a list of books written by `author`
- `__len__` 🟡 — returns how many books are in the library
- `__getitem__(index)` 🟡 — lets you do `library[0]` to get a book by index
- `__iter__` / `__next__` 🔴 — makes the library directly loop-able with `for book in library:`
- `save_to_file(filename)` 🔴 — writes one line per book as `title,author,price,stock`
- `load_from_file(filename, name)` 🔴 — class method that reads the file back into a new `Library`
- `__add__(other)` 🔴 — `library_a + library_b` returns a new `Library` with both libraries' books


```python
import re

class Library:
    def __init__(self, name: str):
        self.name = name
        # TODO 🟡: initialise an empty list of books
        self.books: list[Book] = []
        
    def add_book(self, book: Book):
        # TODO 🟡
        self.books.append(book)

    def total_value(self) -> float:
        # TODO 🟡
        return sum(book.get_price()*book.get_stock() for book in self.books)

    def find_by_author(self, author: str) -> list[Book]:
        # TODO 🟡
        return [book for book in self.books if book.author==author]

    def __len__(self) -> int:
        # TODO 🟡
        return sum(book.get_stock() for book in self.books)

    def __getitem__(self, index: int) -> Book:
        # TODO 🟡
        return self.books[index]

    # ----------- WAIT FILLING THIS UNTIL EXCERCISE 4 --------- 
    def __iter__(self):
        # TODO 🔴
        self.counter = 0
        return self
        # Retorna diretamente o iterador da lista de livros
        #return iter(self.books)
    

    def __next__(self):
        # TODO 🔴
        if self.counter < len(self.books):
            book = self.books[self.counter]
            self.counter+=1
            return book
        else:
            raise StopIteration ("List exausted")

    # ----------- WAIT FILLING THIS UNTIL EXCERCISE 5 --------- 
    def save_to_file(self, filename: str):
        # TODO 🔴
        with open(filename, "w") as file:
            file.writelines(repr(book)+"\n" for book in self.books)



    @classmethod
    def load_from_file(cls, filename: int, name: str="Loaded Library") -> Library:
        # TODO 🔴
        read_pattern = re.compile(r"^Book\(title=(.*?), author=(.*?), price=(\d+), stock=(\d+)\)$")
        library = Library(name)
        with open(filename, "r") as file:
            for book in file:
                match = read_pattern.fullmatch(book.strip())
                if match:
                    title, author, price, stock= match.groups()
                    library.add_book(Book(title=title, author=author, price=float(price), stock=int(stock)))

        return library
    
    # ----------- WAIT FILLING THIS UNTIL EXCERCISE 6 --------- 
    def __add__(self, other: Library) -> Library:
        # TODO 🔴
        library = Library(f"{self.name} + {other.name}")
        library.books = self.books + other.books

        return library
        
```

```python
lib = Library("City Library")
lib.add_book(Book("Dune", "Frank Herbert", 150, 3))
lib.add_book(Book("Project Hail Mary", "Andy Weir", 120, 10))

print(len(lib))              # 2
print(lib[0])                # Dune by Frank Herbert ...
print(lib.total_value())     # 150*3 + 120*10 = 1650
print(lib.find_by_author("Andy Weir"))
print("\n".join(str(book) for book in lib.find_by_author("Andy Weir")))
```
