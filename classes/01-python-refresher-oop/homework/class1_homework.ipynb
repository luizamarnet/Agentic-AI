## Exercise 1: Assistant Request Data

You are given a list of requests sent to an imaginary AI assistant.

### Core tasks

1. Implement `count_requests`.
2. Implement `get_unique_users`.
3. Implement `get_high_priority_requests`.
4. Implement `format_request`.
5. Handle missing dictionary keys safely.

### Stretch tasks

- Count requests by priority.
- Search messages by keyword.
- Sort requests by priority.
- Add type hints and `assert` tests.


````python
requests = [
    {"user": "Ada", "message": "What is Python?", "priority": "normal"},
    {"user": "Grace", "message": "Help! My code crashes.", "priority": "high"},
    {"user": "Ada", "message": "Explain classes.", "priority": "normal"},
    {"user": "Linus", "message": "What is a list?", "priority": "low"},
]


def count_requests_by_priority(requests: list[dict[str,str]]) -> dict[str, int]:
    # TODO: Return the number of requests.
    unique_priorities = {request.get("priority", "Unknown priority") for request in requests}
    count_requests_per_priority = {priority: [request.get("priority", "Unknown priority") for request in requests].count(priority) for priority in unique_priorities}
    return count_requests_per_priority

def count_requests(requests: list[dict[str,str]]) -> int:
    return len(requests)


def get_unique_users(requests: list[dict[str,str]]) -> list[str]:
    # TODO: Return a list of unique users.
    # Keep the order in which they first appear.
    unique_users = list(dict.fromkeys(request.get("user", "Unknown user") for request in requests))
    return unique_users


def get_high_priority_requests(requests: list[dict[str,str]]) -> list[dict[str,str]]:
    # TODO: Return only requests with priority "high".
    return [request for request in requests if request.get("priority")=="high"]


def format_request(request: dict[str,str]) -> str:
    # TODO: Return text in this format:
    # [HIGH] Grace: Help! My code crashes.
    #
    # Use sensible default values if fields are missing.
    priority = request.get("priority", "Unknown priority")
    user = request.get("user", "Unknown user")
    message = request.get("message", "No message.")
    list_texts = f"[{priority.upper()}] {user}: {message}"        

    return list_texts

def search_message_by_keyword(requests: list[dict[str,str]], keyword: str) -> list[dict[str, str]]:

    resquests_keyward = [request for request in requests if (msg := request.get("message")) and keyword.lower() in msg.lower()]

    if not resquests_keyward:
        return [{"No messages with keyward": keyword}]
    else:
        return [{request.get("user", "Unknown user"): request["message"]} for request in resquests_keyward]

def sort_requests_by_priority_by_name(requests: list[dict[str,str]], by_name: bool = True) -> list[dict[str, str]]:

    order = {"high": 0, "normal": 1, "low": 2}

    if by_name: 
        return sorted(requests, key=lambda t: (order.get(t.get("priority", "unknown"),99), t.get("user", "")=="" , t.get("user", ""))) 
    else:  
        return sorted(requests, key=lambda t: order.get(t.get("priority", "unknown"),99))

````

````python
print("Number of requests:", count_requests(requests))
print("Unique users:", get_unique_users(requests))
print("High-priority requests:", get_high_priority_requests(requests))

for request in requests:
    print(format_request(request))

incomplete_request = {"user": "Ada", "message": "Hello"}
print(format_request(incomplete_request))

print(count_requests(requests))
````

## Exercise 2: Build a `Message` Class

LLM applications commonly use messages with roles such as:

- `system`
- `user`
- `assistant`

### Core tasks

1. Store `role` and `content`.
2. Validate roles.
3. Add `display()`.
4. Add `word_count()`.
5. Add `is_empty()`.

### Stretch tasks

- Add `contains_keyword(keyword)`.
- Implement `__str__`.
- Normalize roles with `.strip().lower()`.
- Add metadata or a timestamp.


````python
from datetime import datetime


class Message:
    ALLOWED_ROLES = frozenset({"system", "user", "assistant"})

    def __init__(self, role: str, content: str):
        # TODO: Validate the role.

        role = role.strip().lower()
        if role not in Message.ALLOWED_ROLES:
            raise ValueError(
                f"Invalid role: {role}. Allowed roles are: {Message.ALLOWED_ROLES}"
            )
        # TODO: Store role and content.
        self.role = role
        self.content = content
        self.date_of_creation = datetime.now().astimezone().strftime("%Y-%m-%d %H:%M:%S - %z")
        

    def display(self) -> str:
        # TODO: Return text such as:
        # USER: Explain Python classes.
        return f"[{self.date_of_creation}] {self.role.upper()}: {self.content}"

    def word_count(self) -> int:
        # TODO: Return the number of words in content.
        return len(self.content.split())

    def is_empty(self) -> bool:
        # TODO: Return True if content is empty or whitespace only.
        return not bool(self.content.strip())

    def contains_keyword(self, keyword: str) -> bool:

        return bool(keyword.lower() in self.content.lower())

    def __str__(self) -> str:
        return self.display()
````

````python
message = Message("user", "Explain Python classes.")

print(message.display())
print("Word count:", message.word_count())
print("Is empty:", message.is_empty())
print(message)

assert message.role == "user"  # normalized
#assert str(message) == "USER: Explain Python classes."
assert message.contains_keyword("python") is True
assert message.contains_keyword("java") is False

# Try after role validation is implemented:
#Message("developer", "Hello")

try:
    Message("developer", "Hello")
    assert False, "should have raised ValueError"
except ValueError:
    pass

try:
    Message("developer", "Hello")
except ValueError as e:
    print(e)
````

## Exercise 3: Build a `Conversation` Class

A conversation contains multiple `Message` objects.

### Core tasks

1. Store messages in a list.
2. Implement `add_message(message)`.
3. Implement `message_count()`.
4. Implement `transcript()`.
5. Implement `latest_message()`.
6. Implement `count_by_role(role)`.

### Stretch tasks

- Add `get_messages_by_role(role)`.
- Add `clear()`.
- Limit the maximum number of stored messages.
- Validate that only `Message` objects can be added.


````python
class Conversation:
    def __init__(self, max_messages: int = 10):
        # TODO: Create an empty list for messages.
        self.messages: list[Message] = []
        self.max_messages = max_messages

    def add_message(self, message: Message) -> None:
        # TODO: Add a Message object.
        if not isinstance(message, Message):  # type: ignore
            raise TypeError(f"Error: object of type {type(message)}. Only objects of type Message are accepted") 
        if self.message_count() >= self.max_messages:
            self.messages.pop(0)
        self.messages.append(message)

    def message_count(self) -> int:
        # TODO: Return the number of messages.
        return len(self.messages)

    def transcript(self) -> str:
        # TODO: Return all messages on separate lines.
        return ("\n").join(message.display() for message in self.messages)

    def latest_message(self) -> Message: 
        # TODO: Return the newest message.
        # Decide what should happen for an empty conversation.
        return "Missing message" if self.messages[-1].is_empty() else self.messages[-1]

    def count_by_role(self, role: str) -> int:
        # TODO: Return the number of messages with this role.
        return [message.role for message in self.messages].count(role) 

    def clear(self):
        self.messages = []

    def get_message_by_role(self, role: str) -> list[str]:
        return [message for message in self.messages if message.role==role]

    
````

````python
conversation = Conversation()

conversation.add_message(Message("user", "What is OOP?"))
conversation.add_message(
    Message("assistant", "OOP organizes programs using objects.")
)

print("Message count:", conversation.message_count())
print("Latest message:", conversation.latest_message().display())
print("User messages:", conversation.count_by_role("user"))
print()
print(conversation.transcript())

c = Conversation(max_messages=2)
c.add_message(Message("user", "one"))
c.add_message(Message("user", "two"))
c.add_message(Message("user", "three"))  # should push out "one"
assert c.message_count() == 2
assert [m.content for m in c.messages] == ["two", "three"]

try:
    c.add_message("not a message")
    assert False, "should have raised TypeError"
except TypeError as e:
    print(e)

c.clear()
assert c.message_count() == 0
print("Exercise 3 bonus tests passed!")

````

````python
from collections.abc import Callable
from types import MappingProxyType
from typing import Final


class ToolResult:
    def __init__(self, status:bool, value:str):
        # TODO: Call the parent constructor.
        self.status = status
        self.value = value

    def __str__(self):
        status = "[Ok]" if self.status else "[Error]"
        return f"{status} {self.value}" 


class Tool:
    def __init__(self, name: str, description: str):
        self.name = name
        self.description = description

    def run(self, input_text: str) -> ToolResult:
        raise NotImplementedError("Each tool must implement run().")


class CalculatorTool(Tool):
    
    _OPERATIONS: Final[dict[str, Callable[[float,float],float]]] = {
        "+": lambda a, b: a + b,
        "-": lambda a, b: a - b,
        "*": lambda a, b: a * b,
        "/": lambda a, b: a / b,
    }
    VALID_OPERATORS: Final[frozenset[str]] = frozenset(_OPERATIONS.keys())

    def __init__(self):
        # TODO: Call the parent constructor.
        super().__init__("Calculator", f"Performs the operations {CalculatorTool.VALID_OPERATORS} between two numbers")

    def run(self, input_text: str) -> ToolResult:
        # TODO: Support input such as "4 + 7".
        # TODO: Return a helpful error for invalid input.
        try:
            num1, operator, num2 = input_text.split()
        except ValueError:
            return ToolResult(False, "Wrong format. Expected <number_1> <opertor> <number_2>")

        if operator not in CalculatorTool.VALID_OPERATORS:
            return ToolResult(False, f"Invalide operator. Expected one of {self.VALID_OPERATORS}.")

        try:
            float_num1, float_num2 = float(num1), float(num2)
            return ToolResult(True, f"{input_text}: {CalculatorTool._OPERATIONS[operator](float_num1,float_num2)}")
        except ValueError:
            return ToolResult(
                False, 
                "Invalid numbers. Both numbers need to be valid floats or integers."
                ) 
        except ZeroDivisionError:
            return ToolResult(False, "Invalid number. The second number can not be zero")
                
        

        '''if not num1.replace(".", "", 1).lstrip("-").isdigit() or not num2.replace(".", "", 1).lstrip("-").isdigit():
            return ToolResult(
                False, 
                "Invalid object. The first and the last elements need to be numbers writen in number format. Please use a dot for decimal numbers."
                ) '''      
        
        '''if float(num2)==0 and operator=="/":
            return ToolResult(False, "Invalid number. The second number can not be zero")'''       
    


class TextStatisticsTool(Tool):
    def __init__(self):
        # TODO: Call the parent constructor.
        super().__init__("Text Statistics Tool", "Counts caracters and words")

    def run(self, input_text: str) -> ToolResult:
        # TODO: Return character and word counts.
        n_words = len(input_text.split())
        n_caracters = len(input_text)

        return ToolResult(True, f"Statement '{input_text}' has {n_words} words and {n_caracters} caracters")

   
class MockWeatherTool(Tool):
    FAKE_WEATHERS = MappingProxyType({"aarhus": "14°C, cloudy", "copenhagen": "13°C, windy"})
    def __init__(self):
        super().__init__("Weather mocker", "Generates fake weather")

    def run(self, input_text: str) -> str:
        return MockWeatherTool.FAKE_WEATHERS.get(input_text.strip().lower(), f"unknown weather for {input_text}")

````

## Exercise 5: Build a Rule-Based Assistant

This is **not** an LLM. It uses rules to choose a tool.

### Expected behaviour

```text
User: calculate 8 + 12
Assistant: Calculator result: 20
```

### Core tasks

1. Store user input in the conversation.
2. Implement `get_tool(name)`.
3. Route `calculate ` commands to the calculator.
4. Route `count words in ` commands to the statistics tool.
5. Return a help message for unknown input.
6. Store assistant responses in the conversation.

### Stretch tasks

- Add `help`.
- Add `history`.
- Match commands case-insensitively.
- Add `calc ` as an alias.


````python
class SimpleAssistant:
    """
    Use modes:
        'calculate <a> <operator> <b>'
        count words in <phrase>
    Example:
        calculate 2 + 5
        count words in what is your name?
    """
    def __init__(self, tools: list[CalculatorTool|MockWeatherTool|TextStatisticsTool]):
        self.tools = tools
        self.conversation = Conversation()

    def help(self) -> str:
        return self.__doc__ # type: ignore

    def history(self) -> str :
        return self.conversation.transcript()

    def get_tool(self, name:str) -> CalculatorTool|MockWeatherTool|TextStatisticsTool:
        # TODO: Return the matching tool or None.
        for tool in self.tools:
            if tool.name.lower() == name.lower():
                return tool   
        return "unknown tool" # type: ignore
            
    def respond(self, user_input: str):
        # TODO: Store a user Message.
        self.conversation.add_message(Message("user", user_input))
        # TODO: Route tool requests.
        if "chat history" in user_input.lower():
            response = self.history()
            self.conversation.add_message(Message("assistant", "user asked for chat history"))
            return response
        if user_input.lower().startswith("help"):
            response = self.help()
            self.conversation.add_message(Message("assistant", response))
            return response
        if user_input.lower().startswith("calculate ") or user_input.lower().startswith("calc "):
            name = "Calculator"
            value = user_input.lower().split("calculate ")[-1] if user_input.lower().startswith("calculate ") else user_input.lower().split("calc ")[-1] 
        elif user_input.lower().startswith("count words in "):
            name = "Text Statistics Tool"
            value = user_input.lower().split("count words in ")[-1]
        else:
            response = "Not known tool. Try ask for <calculate> <expression> or <count words in> <phrase>"
            self.conversation.add_message(Message("assistant", response))
            return response
        tool = self.get_tool(name)
        response = tool.run(value)
        # TODO: Store an assistant Message.
        self.conversation.add_message(Message("assistant", response))
        # TODO: Return the response text.
        return response
````

````python
assistant = SimpleAssistant(
    tools=[
        CalculatorTool(),
        TextStatisticsTool(),
    ]
)

print(assistant.respond("calculate 8 + 12"))
print(assistant.respond("Calc 8 + 2"))
print(assistant.respond("Calc 8 / 0"))
print(assistant.respond("count words in Agentic systems use tools"))
print(assistant.respond("Hello there"))
print(assistant.respond("Help me on how to use this assistant"))
print(assistant.respond("what is our Chat history"))

print("-"*10)
print("Conversation transcript:")
print(assistant.conversation.transcript())

print("-"*10)

print("\nDOC: \n",assistant.help())
#print(help(SimpleAssistant))

print(assistant.history())
````
