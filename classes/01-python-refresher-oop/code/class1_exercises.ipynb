### A1 🟢 Core — Variables & control flow: a tiny decision-maker

Later in this course you'll build AI *agents* that look at some input and decide what to do. <br>
Let's warm up with the simplest possible version of that idea: *a function that looks at a number and decides what "action" to take.*

Let's write a function `classify_temperature(celsius)` that returns:
- `"freezing"` if celsius <= 0
- `"cold"` if 0 < celsius <= 15
- `"mild"` if 15 < celsius <= 25
- `"hot"` if celsius > 25


````python
def classify_temperature(celsius: float) -> str:
    # TODO: implement this
    if celsius<=0:
        return "freezing"
    elif 0<celsius<=15:
        return "cold"
    elif 15<celsius<=25:
        return "mild"
    else:
        return "hot"
````

### A2 🟢 Core — Functions: default arguments & `*args`/`**kwargs`

Write a function `build_greeting(name, greeting="Hello", **details)` that returns a string like:

```
"Hello, Silvia! (role: data steward, city: Aarhus)"
```

- `greeting` should default to `"Hello"` if not provided.
- `**details` can be any number of keyword arguments; join them as `key: value` pairs separated by `", "`, wrapped in parentheses.
- If no extra details are given, just return `"Hello, Silvia!"` with no parentheses.


````python
def build_greeting(name: str, greeting: str="Hello", **details: dict[str,str]):
    # TODO: implement this
    if not details:
        return (f"{greeting}, {name}!")
    else:
        aux_details = (f"{key}: {value}" for key, value in details.items())
        print(type(aux_details))
        aux_string = next(aux_details)
        for detail in aux_details:
            aux_string = aux_string + ", " + detail 
        return (f"{greeting}, {name}! ({aux_string})")
````

````python
# Self-check
assert build_greeting("Silvia", role="data steward", city="Aarhus") == \
    "Hello, Silvia! (role: data steward, city: Aarhus)"
assert build_greeting("Azim") == "Hello, Azim!"
assert build_greeting("Azim", greeting="Hi") == "Hi, Azim!"
print("A2 passed!")
````

### A3 🟢 Core — Data structures: filtering and sorting a list of dicts

You're given a list of "tasks" that a future agent might need to work through. Each task is a dictionary with `name`, `priority` (1 = highest), and `done` (bool).

Write a function `pending_tasks_by_priority(tasks)` that returns a list of task **names** (strings), containing only tasks where `done` is `False`, sorted by priority ascending (1 first).

Use a **list comprehension** for the filtering step.


````python
tasks: list[dict[str,str|bool|int]] = [
    {"name": "read paper", "priority": 2, "done": False},
    {"name": "write code", "priority": 1, "done": False},
    {"name": "email advisor", "priority": 3, "done": True},
    {"name": "prepare slides", "priority": 1, "done": False},
]

def pending_tasks_by_priority(tasks: list[dict[str,str|bool|int]]) -> list[str]:
    # TODO: implement this using a list comprehension + sorted()
    pending_tasks=[task for task in tasks  if not task["done"]]
    
    if pending_tasks:
        pending_tasks.sort(key=lambda p: p["priority"]) 
        return [task["name"] for task in pending_tasks] # type: ignore
    else: 
        return (["No pending tasks!"])

````

````python
# Self-check
result = pending_tasks_by_priority(tasks)
assert result == ["write code", "prepare slides", "read paper"], result
print("A3 passed!")
````

### A5 🔴 Challenge — Error handling: a flaky "tool call"

Agents call external tools (APIs, calculators, search engines) that can fail. Let's simulate that.

Below is a function `flaky_divide(a, b)` that raises a `ZeroDivisionError` if `b == 0`.

Write a function `safe_call(func, a, b)` that:
- calls `func(a, b)` and returns its result if it succeeds
- catches `ZeroDivisionError` and returns the string `"error: division by zero"` instead of crashing
- catches any other `Exception` and returns `f"error: {type(e).__name__}"`


````python
from collections.abc import Callable


def flaky_divide(a: float, b:float) -> float:
    return a / b

def safe_call(func: Callable[[float, float], float], a: float, b: float):
    # TODO: implement this with try/except
    try:
        return func(a,b)
    except ZeroDivisionError:
        return "error: division by zero"
    except Exception as e:
        return f"error: {type(e).__name__}"


````

````python
# Self-check
assert safe_call(flaky_divide, 10, 2) == 5.0
assert safe_call(flaky_divide, 10, 0) == "error: division by zero"
assert safe_call(flaky_divide, 10, "a") == "error: TypeError"
print("A5 passed!")
````

## Part B — Introduction to Object-Oriented Programming

So far our "tasks" were just dictionaries floating around. That works, but it gets messy — nothing stops you from forgetting a key, and there's no natural place to put behavior (like "mark this task done").

**Classes** let us bundle data (attributes) and behavior (methods) together. This is the foundation you'll need to build `Agent`, `Tool`, and `Memory` classes later in the course.


### B1 🟢 Core — Your first class

Define a class `Task` with:
- An `__init__(self, name, priority)` that stores `name`, `priority`, and sets `done = False`
- A method `mark_done(self)` that sets `self.done = True`
- A method `__str__(self)` that returns `f"[ ] {name} (priority {priority})"` when not done, or `f"[x] {name} (priority {priority})"` when done

(`__str__` is a special "dunder" method — when you call `print(some_task)` or `str(some_task)`, Python uses whatever `__str__` returns. We'll see more dunder methods next class.)


````python
class Task:
    def __init__(self, name: str, priority:int):
        # TODO: implement this
        self.name = name
        self.priority = priority
        self.done = False

    def mark_done(self):
        self.done = True

    def __str__(self):
        # TODO: implement this
        if self.done:
            return f"[x] {self.name} (priority {self.priority})"
        else:    
            return f"[ ] {self.name} (priority {self.priority})"
        
````

````python
# Self-check
t = Task("write code", 1)
assert t.name == "write code"
assert t.priority == 1
assert t.done is False
assert str(t) == "[ ] write code (priority 1)"
t.mark_done()
assert t.done is True
assert str(t) == "[x] write code (priority 1)"
print("B1 passed!")
````

### B2 🟢 Core — Instance attributes vs. class attributes

Add a **class attribute** `total_created` to `Task` that starts at `0`, and increment it inside `__init__` every time a new `Task` is created. This is shared across *all* instances (unlike `name`/`priority`/`done`, which belong to each individual instance).

Re-define the class below (copy your B1 solution in and extend it).


````python
class Task:
    total_created = 0  # class attribute — shared by ALL instances

    def __init__(self, name: str, priority: int):
        # TODO: set self.name, self.priority, self.done
        self.name = name
        self.priority = priority
        self.done = False
        # TODO: increment Task.total_created
        Task.total_created += 1

    def mark_done(self):
        self.done = True

    def __str__(self):
        # TODO: implement this
        box = "[x]" if self.done else "[ ]"
        return f"[{box}] {self.name} (priority {self.priority})"
        
````

### B3 🟡 Stretch — A class that *uses* another class

Define a class `TaskList` that:
- `__init__(self)` sets `self.tasks = []`
- `add(self, task)` appends a `Task` object to `self.tasks`
- `pending(self)` returns a list of `Task` objects where `done` is `False`, sorted by priority
- `__len__(self)` returns the number of tasks in the list (so `len(my_list)` works)

This mirrors what you did in A3, but now the *behavior* lives inside the object instead of being a free-floating function.


````python
class TaskList:
    def __init__(self):
        # TODO
        self.tasks: list[Task] = []

    def add(self, task: Task):
        # TODO
        self.tasks.append(task)

    def pending(self):
        # TODO
        pending_tasks = [task for task in self.tasks if not task.done]
        pending_tasks.sort(key = lambda s : s.priority)
        return pending_tasks

    def __len__(self):
        # TODO
        return len(self.tasks)
````

````python
# Self-check
tl = TaskList()
tl.add(Task("write code", 1))
tl.add(Task("read paper", 2))
tl.add(Task("email advisor", 3))
assert len(tl) == 3
pending_names = [t.name for t in tl.pending()]
assert pending_names == ["write code", "read paper", "email advisor"]
tl.tasks[2].mark_done()
pending_names = [t.name for t in tl.pending()]
assert pending_names == ["write code", "read paper"]
print("B3 passed!")
````

### B4 🔴 Challenge — A (very) minimal Agent

This is a sneak preview of where the course is going. An "agent" — in the simplest possible sense — is just an object that:
1. has a **goal**
2. has some **state** (e.g. a to-do list)
3. has a method that looks at its state and **decides what to do next**

Define a class `SimpleAgent` with:
- `__init__(self, name, task_list)` storing `self.name` and `self.task_list` (a `TaskList`)
- a method `act(self)` that:
  - if there are no pending tasks, returns `f"{name}: nothing to do!"`
  - otherwise, picks the **highest-priority pending task** (lowest number), marks it done, and returns `f"{name}: working on '{task.name}'"`

Try calling `.act()` repeatedly on an agent with a few tasks and watch it work through them one at a time.


````python
class SimpleAgent:
    def __init__(self, name: str, task_list: TaskList):
        # TODO
        self.name = name
        self.task_list = task_list

    def act(self):
        # TODO
        pending_task = self.task_list.pending()
        if not pending_task:
            return f"{self.name}: nothing to do!"
        task = pending_task[0]
        task.mark_done()
        return f"{self.name}: working on '{task.name}'"
````

````python
# Self-check
tl = TaskList()
tl.add(Task("write code", 1))
tl.add(Task("read paper", 2))
agent = SimpleAgent("Agent Smith", tl)

r1 = agent.act()
assert r1 == "Agent Smith: working on 'write code'", r1
r2 = agent.act()
assert r2 == "Agent Smith: working on 'read paper'", r2
r3 = agent.act()
assert r3 == "Agent Smith: nothing to do!", r3
print("B4 passed! You just built your first (extremely simple) agent.")
````
